```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("D:\\SEMESTER 6\\Studi Independen\\EDA DATA KATEGORISASI\\hasil_preprocessing_newest.csv")
df.info()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# cek label yang ada
print(df["label"].value_counts())

# fungsi membuat word cloud
def plot_wordcloud(label):
    text = " ".join(df[df["label"] == label]["clean_text"].astype(str))

    if len(text.strip()) == 0:
        print(f"Tidak ada teks untuk label {label}")
        return

    wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        max_words=100,
        collocations=False
    ).generate(text)

    plt.figure(figsize=(12,6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title(f"Word Cloud - Label {label}", fontsize=14)
    plt.show()

# tampilkan untuk setiap label
for label in sorted(df["label"].unique()):
    plot_wordcloud(label)
```
